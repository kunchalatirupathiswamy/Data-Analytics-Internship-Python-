```python
#Level-3
#Tsk-1
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("/content/Dataset .csv")
print(df["Rating text"].value_counts())

#average length of reviews and explore
df["Review Length"] = df["Rating text"].str.len()
print("Average Length:", df["Review Length"].mean())
print(df.groupby("Rating text")["Aggregate rating"].mean())
```

```python
#Level-3
#Task-2
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("/content/Dataset .csv")
print("Highest Votes:")
print(df.nlargest( 5, "Votes")[["Restaurant Name", "Votes"]])
print("Lowest Votes:")
print(df.nsmallest( 5, "Votes")[["Restaurant Name", "Votes"]])

#correlation
print("Correlation:")
print(df["Votes"].corr(df["Aggregate rating"]))
```

```python
#Level-3
#Task-3
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("/content/Dataset .csv")
print(pd.crosstab(df["Price range"], df["Has Online delivery"]))
print(pd.crosstab(df["Price range"], df["Has Table booking"]))
print(df.groupby("Price range")["Aggregate rating"].mean())
```
