```python
#Level-2
#Task-1
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("/content/Dataset .csv")
rating = df[df["Aggregate rating"] > 0]
print("Average Rating:", rating["Aggregate rating"].mean())
print("Highest Rating:", rating["Aggregate rating"].max())
print("Lowest Rating:", rating["Aggregate rating"].min())
print("Median Rating:", rating["Aggregate rating"].median())

# Average number of votes
Votes = df["Votes"].mean()
print("Average No.of Votes:",Votes)
```

```python
#Level-2
#Task-2
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("/content/Dataset .csv")

# Remove empty cuisine values
cuisine = df.dropna(subset=["Cuisines"])

# Most common cuisine combinations
print(cuisine["Cuisines"].value_counts().head(10))

# Higher rating for each cuisine combination
Higher = df.groupby('Cuisines')['Aggregate rating'].mean().sort_values(ascending=False)
print(Higher.head(10))
```

```python
#Level-2
#Task-3
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("/content/Dataset .csv")
geo = df.dropna(subset=['Longitude', 'Latitude'])
plt.scatter(geo['Longitude'], geo['Latitude'], alpha=0.5)
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.title('location of Restaurants')
plt.show()


# restaurants in specific areas.
clusters = df['City'].value_counts().head(10)
print("Top restaurant clusters:")
print(clusters)
```

```python
#Level-2
#Task-4
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("/content/Dataset .csv")
# Count restaurants by name
chains = df['Restaurant Name'].value_counts()
print("Restaurant Chains:")
print(chains.head(20))

# restaurant chain analysis
chains = df.groupby('Restaurant Name').agg(Outlets=('Restaurant Name', 'count'),
Avg_Rating=('Aggregate rating', 'mean'),Total_Votes=('Votes', 'sum'))

# sort by popularity
chains = chains.sort_values('Total_Votes', ascending=False)
print(chains.head(10))
```
