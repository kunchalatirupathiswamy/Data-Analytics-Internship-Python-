```python
#Level-1
#Task-2
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("/content/Dataset .csv")
city_count = df["City"]
print(city_count.head(1))

#Average rating for each city
avg_rating = df.groupby("City")["Aggregate rating"].mean()
print("Average rating:",avg_rating)

#City with highest average rating
highest_rating = df.groupby("City")["Aggregate rating"].mean().idxmax()
print("higest Average rating:",highest_rating)
```

```python
#level-1
#Task-3
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("/content/Dataset .csv")

#Price Range
price_range = df["Price range"].value_counts().sort_index()
print(price_range)

#Bar chart
plt.bar(price_range.index, price_range.values)
plt.xlabel("Price range")
plt.ylabel("Price_range")
plt.title("Number of Resturents")
plt.show()

# Percentage of restaurants in each price range
percentage = price_range / price_range.sum() * 100
print(percentage)
```

```python
#Level-1
#Task-4
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("/content/Dataset .csv")

# Percentage of restaurants offering online delivery
online_delivery = df["Has Online delivery"].value_counts()
print("Percentageof Online Delivery:",online_delivery/len(df)*100)

# # Average rating with/without delivery
average_rating = df.groupby("Has Online delivery")["Aggregate rating"].mean()
print("Average rating:",average_rating)
```
